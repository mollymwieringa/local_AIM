The objective of this notebook is to produce restart files with the desired sea ice and snow thicknes initial conditions for each simulation.

To accomplish this goal, we will need:
- restart files for each initialization from the last year of spinup run
- initial conditions for each initialization date
- code from Icepack on how to distribution hi/hs_init across the categories for slab and ITD grid cells 

List of CB initialization dates:
- 2024/2025 CONTROL + FLOODING SIMULATIONS: 2024-11-26
    - R3: hi=0.377, hs=0.045
    - R4: hi=0.5, hs=0.033
    - R7:  hi=0.375, hs=0.03
- 2024/2025 POSTFLOOD SIMULATIONS (3DIC): 
    - R3: 2025-02-16, hi=1.35, hs=0.0
    - R4: 2025-02-17, hi=1.455, hs=0.0
    - R7A: 2025-01-21, hi=1.121, hs=0.0
    - R7B: 2025-02-17, hi=1.443, hs=0.0

List of SV initialization dates:
- 2024 CONTROL + FLOODING SIMULATIONS: 2024-04-01
    - A/SIMBA 4: hi = 0.7, hs=0.1

```python
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Store Cambridge Bay initial conditions in a dictionary
hi_initial_conditions = {'R3':0.422, 'R4': 0.5, 'R7': 0.42}
hs_initial_conditions = {'R3':0.045, 'R4': 0.033, 'R7': 0.03}

ic_year = '2024'
ic_month = '11'
ic_day = '26'
```

```python
# Store Svalbard initial conditions in a dictionary
hi_initial_conditions = {'A': 0.83, 'B': 0.84, 'C': 0.84, 'D': 0.83}  # A = 0.83, 0.17, B=0.84, 0.18 C=0.84, 0.17 D = 0.83, 0.13
hs_initial_conditions = {'A': 0.17, 'B': 0.18, 'C': 0.17, 'D': 0.13}

ic_year = '2024'
ic_month = '04'
ic_day = '01'
```

#### CONTROL / FLOODING

```python
# Cambridge Bay
site='R7'
restart_file = '/Users/mollyw/Desktop/ARISE-AIM/icepack-dirs/runs/CMBDGBY_SPINUP/restart/iced.2027-'+ic_month+'-'+ic_day+'-00000.nc'
right_date_file = '/Users/mollyw/Desktop/ARISE-AIM/icepack-dirs/runs/CMBDGBY_SPINUP/restart/iced.'+ic_year+'-'+ic_month+'-'+ic_day+'-00000.nc'
restart_ds = xr.open_dataset(restart_file)
attributes = xr.open_dataset(right_date_file).attrs
restart_ds.attrs = attributes

restart_ds['vicen'][1] = hi_initial_conditions[site] * restart_ds['aicen'][1]
restart_ds['vicen'][2] = hi_initial_conditions[site] * restart_ds['aicen'][2]

restart_ds['vsnon'][1] = hs_initial_conditions[site] * restart_ds['aicen'][1]
restart_ds['vsnon'][2] = hs_initial_conditions[site] * restart_ds['aicen'][2]

restart_ds.to_netcdf('/Users/mollyw/Desktop/ARISE-AIM/Model/restarts/'+site+'_iced.'+ic_year+'-'+ic_month+'-'+ic_day+'-00000.nc')
```

```python
# Svalbard 
site='D'
restart_file = '/Users/mollyw/Desktop/ARISE-AIM/icepack-dirs/runs/SVALBARD_SPINUP/restart/iced.2025-'+ic_month+'-'+ic_day+'-00000.nc'
right_date_file = '/Users/mollyw/Desktop/ARISE-AIM/icepack-dirs/runs/SVALBARD_SPINUP/restart/iced.'+ic_year+'-'+ic_month+'-'+ic_day+'-00000.nc'
restart_ds = xr.open_dataset(restart_file)
attributes = xr.open_dataset(right_date_file).attrs
restart_ds.attrs = attributes

restart_ds['vicen'][1] = hi_initial_conditions[site] * restart_ds['aicen'][1]
restart_ds['vicen'][2] = hi_initial_conditions[site] * restart_ds['aicen'][2]

restart_ds['vsnon'][1] = hs_initial_conditions[site] * restart_ds['aicen'][1]
restart_ds['vsnon'][2] = hs_initial_conditions[site] * restart_ds['aicen'][2]

restart_ds.to_netcdf('/Users/mollyw/Desktop/ARISE-AIM/Model/restarts/Svalbard_'+site+'_iced.'+ic_year+'-'+ic_month+'-'+ic_day+'-00000.nc')
```

This script performs the simple task of localizing and artificially extending the CERES data record through June of 2025.

```python
import xarray as xr
import pandas as pd
import matplotlib.pyplot as plt
```

```python
station = 'Svalbard'
input_data_path = '/Users/mollyw/Desktop/ARISE-AIM/Data/raw/CERES/'
output_data_path = '/Users/mollyw/Desktop/ARISE-AIM/Data/postprocessed/atm/'
file = 'CERES_SYN1deg-1H_Terra-Aqua-NOAA20_Ed4.2_Subset_20220101-20251130.nc'
ds = xr.open_dataset(input_data_path + file)
```

```python
if station == 'CambridgeBay':
    lat = 69.5
    lon = -105.5
elif station == 'PistoletBay':
    lat = 51.5
    lon = -55.5
elif station == 'Svalbard':
    lat = 78.5
    lon = 15.5
else:
    raise ValueError('Station not recognized, no lat,lon to provide!')
```

```python
local_ds = ds.sel({'lat': lat, 'lon': lon}, method='nearest')
local_ds = xr.concat([local_ds.sel(time=slice('2024-01-01', '2025-12-01')),
                      local_ds.sel(time=slice('2024-12-01', '2025-02-12'))],
                      dim='time')
new_time = pd.date_range('2024-01-01', '2026-02-13', freq='1h')
local_ds['time'] = new_time[:-1]
```

```python
local_ds.to_netcdf(output_data_path + station + '_CERES.nc')
```
